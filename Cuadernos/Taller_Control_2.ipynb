## Reglas que hacen justo y verificable el taller

1. **Mismo corpus:** BM25, E5 y RRF deben operar sobre los mismos 200 documentos. Si cada motor recibe otro conjunto, la comparación carece de valor.
2. **Consultas personales:** una intención profesional se asigna determinísticamente por los códigos de la pareja. Recibirás una redacción literal y otra en paráfrasis.
3. **Relevancia externa:** los documentos «positivos» provienen de un conjunto pedagógico curado, no del score que produce el modelo. Hay documentos discutibles: eso es parte del análisis.
4. **Fuente real, no simulación:** Atlas solo cuenta como Atlas cuando se ejecuta una consulta remota. El vector de NumPy es real, pero se ejecuta localmente.
5. **Multimedia con procedencia:** el PNG es un gráfico de tu ranking. Los diez WAV son grabaciones reales de MINDS14. Son fuentes distintas que NO se presentan como contratos con audio.
6. **Evaluación parcial:** cada control es independiente. El corrector no vuelve a calificar ingesta SECOP, Cassandra, Neo4j o ZIP como habilidades académicas.
7. **Aprender de fallar:** una predicción equivocada no da cero. Una función que no completaste sí deja sin evidencia ese control, pero puedes corregirla y reejecutarla.

### E1.1 · Una palabra rara puede importar más que una común

**Frecuencia en un documento (TF):** cuántas veces aparece una palabra dentro del documento. **Frecuencia documental (DF):** en cuántos documentos aparece, al menos una vez. **IDF:** una medida del peso relativo de las palabras menos comunes.

La forma que usa nuestro BM25 es `IDF = ln(1 + (N - DF + 0.5)/(DF + 0.5))`, donde N es el número de documentos. Observa que **la misma palabra no obtiene el mismo peso en todos los corpus**.

**Tu tarea (COMPLETA):** escribir una sola línea de Python para contar el número de documentos que contienen un término de la consulta de tu equipo. Recuerda convertir las palabras de cada documento en un **conjunto**, para que una palabra repetida cinco veces no cuente como cinco documentos.

**Ejemplo resuelto:** si «vías» aparece en 3 de 10 documentos, entonces DF=3; si se repite siete veces en uno de esos documentos, DF sigue siendo 3.

**Evidencia:** `E1/laboratorio_bm25.json`, con tu término, su DF real, N=200 e IDF.

```python
# E1.1 · COMPLETA · Escribe el conteo de frecuencia documental
import math
def frecuencia_documental_equipo(coleccion,termino):
    # COMPLETA SOLO LA SIGUIENTE LÍNEA.
    # Pista: usa el ejemplo manual anterior y verifica con el assert.
    return sum(1 for d in coleccion if termino in set(core.tokens(core.texto(d))))

TERMINO_LAB=core.tokens(TEMA["literal"])[0]
DF_LAB=frecuencia_documental_equipo(DOCUMENTOS,TERMINO_LAB)
assert isinstance(DF_LAB,int) and 0<=DF_LAB<=len(DOCUMENTOS)
IDF_LAB=math.log(1+(len(DOCUMENTOS)-DF_LAB+.5)/(DF_LAB+.5))
print(f"Palabra {TERMINO_LAB!r}: N={len(DOCUMENTOS)}, DF={DF_LAB}, IDF={IDF_LAB:.4f}")
core.escribir(TRABAJO/"E1/laboratorio_bm25.json",{"termino":TERMINO_LAB,"df":DF_LAB,
 "idf":round(IDF_LAB,8),"n":len(DOCUMENTOS)})
```

**Cómo se lee:** DF indica en cuántos expedientes aparece el término; IDF tiende a bajar cuando el término es muy común. **Qué nos dice:** por qué una palabra puede discriminar mejor que otra. **Qué NO permite concluir:** que una palabra rara haga relevante un documento. **Error común:** sumar apariciones totales en lugar de documentos distintos.

### E2.0 · De palabras a geometría: resuelve primero un ejemplo pequeño

Un **embedding** es un vector calculado a partir de un texto usando un modelo entrenado. No es una bolsa de palabras ni una lista de etiquetas humanas. Un índice vectorial ayuda a buscar vectores cercanos; la **similitud coseno** mide una relación geométrica y no certifica relevancia.

**Ejemplo resuelto, dos dimensiones:** A=(3,4), B=(6,8). El producto punto es `3×6 + 4×8 = 50`. Las normas valen `5` y `10`. Luego `cos(A,B)=50/(5×10)=1`. Ambos tienen la misma dirección aunque magnitudes diferentes. Si C=(−4,3), entonces `cos(A,C)=0`: son perpendiculares.

**Cuidado:** el modelo real de la sesión devuelve **384 dimensiones**; aquí usamos dos para explicar el algoritmo. La prueba de dos dimensiones no reemplaza los embeddings E5. Antes de abrir los pesos del modelo, completa una función matemática que puedas defender sin IA.

```python
# E2.0 · EJECUTA · Producto punto, normas y un contraejemplo
import numpy as np
A=np.array([3.,4.]); B=np.array([6.,8.]); C=np.array([-4.,3.])
print("Producto punto A·B:",float(np.dot(A,B)))
print("Norma de A:",float(np.linalg.norm(A)))
print("Norma de B:",float(np.linalg.norm(B)))
print("¿Por qué A y B apuntan en la misma dirección aunque A ≠ B?")
print("A·C=",float(np.dot(A,C)),"; valor esperado de coseno A,C=0")
```

### E2.1 · COMPLETA · El coseno sin bibliotecas especializadas

**Tu función debe aceptar dos vectores NumPy** y devolver un número entre −1 y 1 cuando los vectores son válidos. Para hacerlo, calcula producto punto dividido por producto de normas.

**Pista:** `np.dot(a,b) / (np.linalg.norm(a) * np.linalg.norm(b))`. No basta con devolver 1 o 0: tu código debe funcionar también en vectores con otra dirección.

**La trampa habitual:** para a=(0,0), el coseno no está definido, pues la norma es cero. El motor del taller rechaza ese vector en lugar de inventar un resultado. En esta práctica los vectores A, B, C son no nulos.

**La evidencia nueva:** `E2/laboratorio_coseno.json`. No se concede crédito por copiar directamente el resultado de una calculadora si no se ejecuta el procedimiento.

```python
# E2.1a · COMPLETA · Implementa una fórmula de coseno
def coseno_mini(a,b):
    # COMPLETA: usa np.dot y np.linalg.norm, devuelve un float.
    return float(np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b)))
COS_AB=coseno_mini(A,B)
COS_AC=coseno_mini(A,C)
assert np.isclose(COS_AB,1.0,atol=1e-5) and np.isclose(COS_AC,0.0,atol=1e-5)
print("cos(A,B):",COS_AB,"; cos(A,C):",COS_AC)
core.escribir(TRABAJO/"E2/laboratorio_coseno.json",{"cos_ab":round(float(COS_AB),6),
 "cos_ac":round(float(COS_AC),6),"formula":"producto_punto_dividido_por_normas"})
```

**Cómo se lee:** coseno 1 indica la misma dirección; 0 indica direcciones ortogonales; −1 indica direcciones opuestas. **Qué nos dice:** que podemos comparar representaciones numéricas aunque no tengan la misma magnitud. **Qué NO permite concluir:** que coseno 1 signifique identidad textual, verdad o pertinencia jurídica. **Error común:** usar distancia euclidiana y etiquetarla como coseno.

## E3 · ¿Cuál buscador recupera documentos útiles? — 10 puntos

El score del motor NO es una calificación humana. La **pertinencia** se evalúa comparando el ranking con un conjunto de documentos relevantes identificado de forma independiente para tu intención.

**Precision@5:** relevantes recuperados entre cinco resultados. **Recall@5:** relevantes recuperados dividido por todos los relevantes conocidos para esa intención en nuestro corpus.

**Ejemplo manual:** de cinco resultados, tres son relevantes; en el corpus se han etiquetado seis relevantes. Precision@5=3/5=0,6 y Recall@5=3/6=0,5. No lo confundas con solapamiento ANN–ENN.

**RRF (Reciprocal Rank Fusion):** combina lugares de dos rankings sin comparar el score de BM25 con el coseno. Cada posición r aporta `1/(60+r)` y se suman los aportes.

**Tu reto:** calcular las métricas y elegir BM25, E5 o RRF a partir de TU resultado, descartando una alternativa y señalando qué evidencia falta.

### E3.0 · ¿Qué significa acertar en una recuperación?

El motor no conoce la respuesta final. Para evaluar si sirve necesitamos una tabla de **relevancia**: documentos que un evaluador considera útiles para una intención concreta. Los cinco temas del TC2 tienen identificadores de referencia curados. Son **juicios pedagógicos para 200 documentos**; no pretenden cubrir el universo completo de compras públicas.

**Ejemplo resuelto:** si el ranking de cinco documentos es A, B, C, D, E y los relevantes de todo el corpus son A, C, F, G, H, I, entonces se recuperan dos relevantes de cinco (**Precision@5 = 2/5 = 0,40**) y dos de seis relevantes existentes (**Recall@5 = 2/6 = 0,3333**). El documento F también puede ser relevante aunque no aparezca en el Top-5.

**Pregunta del experimento:** ¿un motor que devuelve puntuaciones más altas necesariamente obtiene mejor Precision@5? **No**: los scores de BM25 y coseno son escalas diferentes y no equivalen al juicio externo. Espera a ver los datos antes de decidir.

**Dos errores que debes evitar:** (1) dividir Recall@5 entre 5; (2) considerar todo documento con coseno positivo como relevante.

```python
# E3.0 · EJECUTA · Comprueba las métricas a mano con seis positivos
ranking_ejemplo=["A","B","C","D","E"]
positivos_ejemplo={"A","C","F","G","H","I"}
coinciden=[i for i in ranking_ejemplo if i in positivos_ejemplo]
print("Encontrados relevantes:",coinciden)
print("Precision@5 =",len(coinciden),"/",len(ranking_ejemplo),"=",len(coinciden)/5)
print("Recall@5 =",len(coinciden),"/",len(positivos_ejemplo),"=",len(coinciden)/6)
assert len(coinciden)==2
```

### E3.1 · COMPLETA · Calcula Precision@5 sin usar el evaluador listo

La función debe recibir una lista de identificadores y el conjunto de positivos. Usa exclusivamente los **cinco primeros**. Tu trabajo consiste en contar cuántos IDs están en el conjunto de relevantes y dividir por cinco.

**Pista:** `sum(ident in relevantes for ident in ids[:5]) / 5`.

**Importante:** queremos una fórmula que sirva para una lista distinta mañana, no escribir el valor conocido de esta ejecución. Si tu función no depende de `ids` y `relevantes`, no has construido una métrica reutilizable.

**Evidencia:** `E3/laboratorio_precision.json` recalculada sobre tu propio ranking BM25. El corrector comprobará el valor frente a la referencia independiente.

```python
# E3.1b · COMPLETA · La precisión con los IDs de TU pareja
def precision_cinco(ids,relevantes):
    # COMPLETA: cuenta los relevantes recuperados entre los cinco primeros y divide entre 5.
    return sum(ident in relevantes for ident in ids[:5])/5

IDS_BM=[r["id_proceso"] for r in RANK_BM_PARAFRASIS]
PRECISION_MANUAL=precision_cinco(IDS_BM,set(TEMA["relevantes"]))
assert 0<=PRECISION_MANUAL<=1
print("Mi Precision@5 BM25:",PRECISION_MANUAL)
core.escribir(TRABAJO/"E3/laboratorio_precision.json",{
 "top5":IDS_BM[:5],"positivos_totales":len(TEMA["relevantes"]),
 "precision_manual":round(float(PRECISION_MANUAL),4)})
```

**Cómo se lee:** la precisión resume cuántos candidatos entre cinco fueron juzgados útiles. **Qué nos dice:** calidad del conjunto recuperado según nuestros juicios. **Qué NO permite concluir:** que los otros documentos sean irrelevantes para cualquier persona o que el buscador funcionará igual en otra colección. **Error común:** usar el valor del score del motor como etiqueta de referencia.

### E3.2 · ¿Cómo combinar listas sin mezclar escalas?

La **fusión por rango recíproco (RRF)** utiliza el lugar en que cada documento aparece en dos listas. Con constante 60, si el documento X está en la posición 1 de BM25 y posición 3 de E5, su puntuación fusionada es `1/61 + 1/63 ≈ 0,03227`.

Si un documento aparece solo en una de las dos listas, recibe solamente un aporte. Si aparece en ambas, recibe dos. RRF **no sabe** si los rankings tienen errores de interpretación; por eso no reemplaza los juicios de pertinencia.

**Tu tarea:** calcular el aporte de dos posiciones concretas escribiendo una función. Después compáralo con el ranking híbrido calculado sobre el corpus real. No normalices scores BM25 y coseno ni los sumes: sería otra técnica distinta de RRF.

```python
# E3.2b · COMPLETA · Un componente de RRF, desde cero
def rrf_dos_posiciones(posicion_bm25,posicion_e5):
    # COMPLETA: suma dos recíprocos con constante 60; posiciones comienzan en 1.
    return 1/(60+posicion_bm25)+1/(60+posicion_e5)

RRF_EJEMPLO=rrf_dos_posiciones(1,3)
assert abs(RRF_EJEMPLO-(1/61+1/63))<1e-10
print("RRF para posiciones 1 y 3:",round(RRF_EJEMPLO,8))
core.escribir(TRABAJO/"E3/laboratorio_rrf.json",{
 "pos_bm25":1,"pos_e5":3,"k":60,"score":round(float(RRF_EJEMPLO),10)})
```

### E4.0 · Del ranking a los píxeles: por qué es necesario mirar el resultado

Una imagen contiene una cuadrícula de **píxeles**. En RGB, cada píxel tiene componentes rojo, verde y azul. No confundas la cantidad de píxeles con el peso en bytes de un archivo: la compresión puede modificar el tamaño sin modificar las dimensiones.

**Ejemplo:** el color rojo puro (255,0,0) tiene luminancia aproximada 54,213 usando `0,2126R + 0,7152G + 0,0722B`. El verde puro (0,255,0) se percibe como más luminoso por el peso de su canal G.

**COMPLETA una función pequeña** antes de usar la del taller: así podrás interpretar un píxel y verificar la matemática de la luminancia. No basta con saber dónde hacer clic para mostrar el histograma.

**El gráfico que recibirás:** una imagen de 960×520 generada desde TU Top-5 de RRF, con títulos, barras y fuentes. Como contiene texto técnico, reducirlo puede perder información imprescindible.

```python
# E4.0 · COMPLETA · Luminancia aproximada del píxel RGB
def luminancia_rgb(rgb):
    # COMPLETA: devuelve el peso de R, G y B según la fórmula descrita arriba.
    return 0.2126*rgb[0]+0.7152*rgb[1]+0.0722*rgb[2]
LUMINANCIA_ROJO=luminancia_rgb((255,0,0))
LUMINANCIA_VERDE=luminancia_rgb((0,255,0))
assert abs(LUMINANCIA_ROJO-54.213)<.02 and LUMINANCIA_VERDE>LUMINANCIA_ROJO
print("Luminancia rojo puro:",LUMINANCIA_ROJO)
print("Luminancia verde puro:",LUMINANCIA_VERDE)
core.escribir(TRABAJO/"E4/laboratorio_luminancia.json",{
 "rgb_rojo":[255,0,0],"luminancia_rojo":round(float(LUMINANCIA_ROJO),3),
 "rgb_verde":[0,255,0],"luminancia_verde":round(float(LUMINANCIA_VERDE),3)})
```

### E5.1 · COMPLETA · La fórmula RMS en una función que puedas reutilizar

El RMS —*root mean square*, raíz de la media de cuadrados— resume la magnitud de una señal. No es igual al promedio simple: si hay muestras positivas y negativas, el promedio simple puede cancelarlas.

**Ejemplo resuelto:** con `[1,-1,0,0]`, el promedio simple es 0, pero el RMS es `√((1² + (−1)² + 0² + 0²)/4) = √0,5 ≈ 0,7071`.

**Tu tarea:** completar una función con NumPy que calcule la raíz del promedio de cuadrados. **Pista:** `float(np.sqrt(np.mean(np.square(muestras))))`. Debe funcionar con otra secuencia de muestras sin escribir a mano 0,7071.

**Evidencia:** `E5/laboratorio_rms.json`. Esta práctica verifica una habilidad de procesamiento de audio que no formó parte del primer taller.

```python
# E5.1a · COMPLETA · RMS por muestras (no por transcripción)
def rms_manual(muestras):
    # COMPLETA: raíz del promedio de las muestras elevadas al cuadrado.
    return float(np.sqrt(np.mean(np.square(muestras))))

RMS_CUATRO=rms_manual(np.array([1.,-1.,0.,0.]))
assert abs(RMS_CUATRO-0.7071067811865476)<1e-6
print("RMS de cuatro muestras:",round(RMS_CUATRO,6))
core.escribir(TRABAJO/"E5/laboratorio_rms.json",{
 "muestras":[1,-1,0,0],"rms_manual":round(float(RMS_CUATRO),6),
 "media_simple":0.0})
```

**Cómo se lee:** 0,7071 es la magnitud RMS de cuatro muestras de ejemplo, no una emoción de voz. **Qué nos dice:** el promedio simple no mide bien la magnitud de una señal alternante. **Qué NO permite concluir:** qué dijo el hablante ni cuál es su intención. **Error común:** confundir RMS con la media de las amplitudes con signo.
